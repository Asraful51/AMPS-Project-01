```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('2.Student_Mental_Stress_and_Coping_Mechanisms.csv')
```

```python
array = data['Study_Hours_Per_Week'].unique()
array = array * 2
```

```python
# Check array attributes adn types
print("Dimension of the array : ",array.shape)
print("Total number of elements : ",array.size)
print("Number of dimensions: ",array.ndim)
print("Data type of elements: ",array.dtype)
print("Python type:",type(array))
```

```python
print(np.isnan(array))
print(np.nanmean(array))
print(np.nanmedian(array))
print(np.nansum(array))
```

```python
#scakar arithematic on an array
print(array + 1)
print(array - 2)
print(array * 3)
print(array / 4)
```

```python
# Vectorized math functions and constants
print(np.sqrt(array))
print(np.round(array))
print(np.pi)
```

```python
# Element wise atithematic between two arrays
array1 = array
array2 = array * 2
print(array1 + array2)
print(array1 - array2)
print(array1 * array2)
```

```python
studyhour = array

studious = studyhour[studyhour > 70]
print("Study more than 50 hour per week : ",studious)

general = studyhour[(studyhour >= 30) & (studyhour < 50)]
print("General study hor per week : ",general)
```

```python
# Basic statistical operations
print("Sum: ", np.sum(array))
print("Mean: ", np.mean(array))
print("Median: ", np.median(array))
print("Standard Deviation: ", np.std(array))
print("Var", np.var(array))
print("Min: ", np.min(array))
print("Max: ", np.max(array))
print("Argmin", np.argmin(array))
```

```python
# calculate specipic percentiles
print("Percentile: ",np.percentile(array, [0, 25, 50, 75, 100]))
```

```python
#Operation along a specific axis
print(np.sum(array, axis=0))
print(np.mean(array, axis=0))
```
